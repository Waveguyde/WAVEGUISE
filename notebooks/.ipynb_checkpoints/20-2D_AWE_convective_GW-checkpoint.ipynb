## Data Preprocessing
Before the observational data set is ready for the wavelet transform, we must take care of the data boundaries. These non-periodic boundaries can create strong frequency responses and later on fuck up the wavefield segmentation. The first step is to use a least square polynomial 2D fit of max_order and subtract the result. In a second step, the zeroth and first Fourier components are also subtracted. In a final step, a Tukey window is applied to smoothly enforce the signal to zero at the edges. 

```python
orig = ds["Radiance"].isel(time=0,x_along_track=slice(5100,5800),y_cross_track=slice(85,353))
#orig = ds["Radiance"].isel(time=0,x_along_track=slice(5150,5500),y_cross_track=slice(85,353))
orig_copy = orig.copy()
BI_high, BI_low = two.BG_removal(orig_copy.values)
dx, dy = 2000, 2000
nx, ny = orig['x_along_track'].size, orig['y_cross_track'].size
y = np.linspace(0,(ny-1)*dy,ny)*1e-3
x = np.linspace(0,(nx-1)*dx,nx)*1e-3
print('Domain size:', nx, ny)

nmask  = orig.shape
ntap   = int(0.1*np.min(nmask))
signal = utils.smooth_edges(BI_high, ntap, window='tukey')
```

## Wavelet Transform
When we do the juwavelet transform we need to choose the number of scales $js$ and orientations $jt$. The smallest scales is defined as $s_0=4*\delta x$. The number of scales is taken to depend on the data dimensions and since the first Fourier component is already subtracted, we focus here on wavelengths shorter than that. For more details on juwavelet, please have a look at Ungermann and Reichert (2025).

```python
s0 = 4 #smallest scale that we wanna resolve
s1 = 0.5 * min([nx,ny]) #largest scale we wanna resolve (since the first Fourier component is part of the background (BI_low) the largest scale fits twice in the domain)
dj = 1 / 8 # apäter wieder dj=1/8
js = int(1/dj*np.log2(s1/s0))
jt = 21
print('Decomposition shape will be: ', js, jt, nx, ny)

cwt_result = transform.decompose2d(signal,dx,dy,s0*min([dx,dy]),dj,js,jt,aspect=1,nxpad=None,nypad=None,opts={'param': 2*np.pi},mode="scaled",dtype=np.complex128)

print(f"The data resolution is {dx*1e-3:.0f} km and the smallest wavelength you analyze is {cwt_result['period'][0]*1e-3:.0f} km.")
print(f"The data domain is at least {min([nx,ny])*dx*1e-3:.0f} km and the largest wavelength you analyze is {cwt_result['period'][-1]*1e-3:.0f} km.")
```

## Red Noise Filtering
Before we continue with the segmentation, it is very useful to strip off any red noise. This is done by normalizing the wavelet power spectrum using the median and scaled median absolute deviation (sMAD) and setting all wavelet coefficients to zero where the normalized WPS is below a certain threshold (8 seems to be a good number in this application).

```python
def angular_distance(theta1, theta2, period=np.pi):
    """
    Zyklische Winkeldistanz für Orientierungen.

    Für Orientierungen ist period=np.pi oft sinnvoll, weil theta und theta+pi
    dieselbe Achse beschreiben. Falls deine Orientierung gerichtet ist,
    setze period=2*np.pi.
    """
    d = np.abs(theta1 - theta2)
    return np.minimum(d, period - d)


def cluster_wavepackets(
    labels,
    mean_x,
    mean_y,
    mean_wavelength,
    mean_orientation,
    max_spatial_distance,
    max_wavelength_difference,
    max_orientation_difference,
    orientation_period=np.pi,
):
    """
    Clustert Wellenpakete anhand ihrer gemittelten Position und spektralen
    Eigenschaften über einen Nachbarschaftsgraphen.

    Zwei Wellenpakete werden verbunden, wenn alle drei Bedingungen erfüllt sind:
        1. räumliche Distanz <= max_spatial_distance
        2. |Delta Wellenlänge| <= max_wavelength_difference
        3. Winkeldistanz <= max_orientation_difference

    Die finalen Cluster sind die Connected Components dieses Graphen.

    Parameter
    ---------
    labels : np.ndarray
        1-D Array mit den expliziten Label-IDs der Wellenpakete.
        Label 0 sollte hier nicht enthalten sein.
    mean_x, mean_y : np.ndarray
        1-D Arrays mit mittleren x- und y-Positionen pro Label.
    mean_wavelength : np.ndarray
        1-D Array mit mittlerer Wellenlänge pro Label.
    mean_orientation : np.ndarray
        1-D Array mit mittlerer Orientierung pro Label.
    max_spatial_distance : float
        Maximale euklidische Distanz in (x,y), damit zwei Pakete als Nachbarn gelten.
    max_wavelength_difference : float
        Maximale Differenz der mittleren Wellenlänge.
    max_orientation_difference : float
        Maximale Winkeldifferenz.
    orientation_period : float, optional
        Periode der Orientierung:
        - np.pi  : ungerichtete Orientierung (Standard)
        - 2*np.pi: gerichtete Orientierung

    Returns
    -------
    cluster_ids : np.ndarray
        1-D Array gleicher Länge wie labels.
        cluster_ids[i] ist die Cluster-ID des Pakets labels[i].
        Cluster-IDs sind fortlaufend: 1, 2, 3, ...
    label_to_cluster : dict
        Dictionary: label_id -> cluster_id
    cluster_to_labels : dict
        Dictionary: cluster_id -> np.ndarray der zugehörigen Label-IDs
    """
    labels = np.asarray(labels)
    mean_x = np.asarray(mean_x, dtype=float)
    mean_y = np.asarray(mean_y, dtype=float)
    mean_wavelength = np.asarray(mean_wavelength, dtype=float)
    mean_orientation = np.asarray(mean_orientation, dtype=float)

    n = len(labels)

    if n == 0:
        return (
            np.array([], dtype=int),
            {},
            {},
        )

    if np.any(labels == 0):
        raise ValueError("labels sollte nur echte Wellenpaket-Labels enthalten, nicht label=0.")

    if not (
        len(mean_x) == len(mean_y) == len(mean_wavelength) == len(mean_orientation) == n
    ):
        raise ValueError("Alle Eingabearrays müssen die gleiche Länge haben.")

    # Adjazenzliste des Graphen
    neighbors = [[] for _ in range(n)]

    for i in range(n):
        for j in range(i + 1, n):
            dx = mean_x[i] - mean_x[j]
            dy = mean_y[i] - mean_y[j]
            spatial_distance = np.sqrt(dx * dx + dy * dy)

            wavelength_difference = np.abs(mean_wavelength[i] - mean_wavelength[j])
            orientation_difference = angular_distance(
                mean_orientation[i],
                mean_orientation[j],
                period=orientation_period,
            )

            if (
                spatial_distance <= max_spatial_distance
                and wavelength_difference <= max_wavelength_difference
                and orientation_difference <= max_orientation_difference
            ):
                neighbors[i].append(j)
                neighbors[j].append(i)

    # Connected Components bestimmen
    cluster_ids = np.zeros(n, dtype=int)
    current_cluster = 0

    for start in range(n):
        if cluster_ids[start] != 0:
            continue

        current_cluster += 1
        stack = [start]
        cluster_ids[start] = current_cluster

        while stack:
            node = stack.pop()
            for nb in neighbors[node]:
                if cluster_ids[nb] == 0:
                    cluster_ids[nb] = current_cluster
                    stack.append(nb)

    label_to_cluster = {int(label): int(cluster_id) for label, cluster_id in zip(labels, cluster_ids)}

    cluster_to_labels = {}
    for cluster_id in range(1, current_cluster + 1):
        cluster_to_labels[cluster_id] = labels[cluster_ids == cluster_id]

    return cluster_ids, label_to_cluster, cluster_to_labels
```

```python
def find_clusters_in_freq_theta_numba(CWT, segments, eps=0.2, min_samples=2, beta=1):
    WAS   = np.abs(CWT["decomposition"])
    freq  = 2 * np.pi / CWT["period"]
    n_freq = len(freq)
    theta = CWT["theta"]

    labels, amps_seg, freq_seg, theta_seg = two.segments2points_numba(WAS, freq, theta, segments)
    
    ds = np.abs(np.diff(np.log(freq))[0])
    dt = np.diff(theta)[0]
    print('suggested scaling:', ds/dt)
    
    steigung = (ds/dt-ds/dt*beta)/(np.log(freq[0])-np.log(freq[int(n_freq/2)]))
    offset = ds/dt*beta - steigung * np.log(freq[-1])

    pts = np.column_stack([np.log(freq_seg),(steigung * np.log(freq_seg) + offset)*theta_seg])

    cluster_labels = two.dbscan_periodic_theta(pts, eps=eps, min_samples=min_samples, theta_period=np.pi)

    return labels, cluster_labels
```

As mentioned the crucial parameter in the segmentation is the prominence parameter. So, we first need a visual check if the wave packets we would seperate by eye are in fact seperated. For that purpose, we simply plot every single wave packet. It is generally a better idea to have a bit too many wave packets than having too less. 

The dilemma is that the prominence parameter treats every dimension equally. While I want to separate two wave packets from one another in the dimensions of scale and orientation, I don't necessarily want to separate them in the (x,y) dimensions because I know that wave amplitudes can be highly variable and so I actually only want to separate wave packets when the ampltude goes down to zero. In order to fix that, I condense the wave packets down to points and plot them in the regular space of log(k) and theta. 

## Clustering
I want to stitch wave packets together that share scale and orientation. However, if they do not overlap in (x,y) I will keep them as separated wave packets.

```python
pts = []
for idx in range(np.max(WP_labels)):
    mask = amp_seg[idx,:,:] > 0
    a = np.log(np.nanmedian(2*np.pi/lh_seg[idx,:,:][mask]))
    b = np.nanmedian(th_seg[idx,:,:][mask])
    pts.append([a,b])
    
pts = np.asarray(pts)  # your list of [a,b]
cluster_labels = two.dbscan_periodic_theta(pts, eps=0.25, min_samples=2, theta_period=np.pi)

print('Number of Clusters: ', np.max(cluster_labels)+1)
print('Number of Noise Points: ', np.sum(cluster_labels == -1))
```

```python
def plot_AWE(
    lon, lat, data,
    levels=None, cmap="viridis", ax=None,
    *, contourf_kwargs=None
):
    if ax is None:
        raise ValueError("Please pass a Cartopy GeoAxes in `ax`.")

    if contourf_kwargs is None:
        contourf_kwargs = {}

    contour = ax.contourf(
        lon, lat, data,
        levels=levels,
        cmap=cmap,
        transform=ccrs.PlateCarree(),
        **contourf_kwargs
    )
    return contour
```
